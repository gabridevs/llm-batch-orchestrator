```python
from llm_orchestrator.demo import build_demo_batch
from llm_orchestrator.models.config import OrchestratorConfig
from llm_orchestrator.orchestrator import BatchOrchestrator
from llm_orchestrator.strategies import AsyncioStrategy

batch = build_demo_batch(size=8, provider_name='fast-provider')
report = await BatchOrchestrator(
    OrchestratorConfig(max_concurrency=3)
).run(batch, AsyncioStrategy())

report.model_dump(mode='json')
```
